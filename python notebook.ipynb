```python
import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_excel("Cleaned_Data_Set.xlsx")

# Step 1: Dataset overview
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

# Step 2: Basic statistics
print("\nStatistical Summary:")
display(df.describe())

# Step 3: Data quality overview
print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Step 4: Total records
total_records = len(df)

print("\nTotal Records:", total_records)

# Step 5: Total metric value
total_metric_value = df["metric_value"].sum()

print("Total Metric Value:", round(total_metric_value, 2))

# Step 6: Average metric value
average_metric_value = df["metric_value"].mean()

print("Average Metric Value:", round(average_metric_value, 2))

# Step 7: Average freshness
average_freshness = df["freshness_hours"].mean()

print("Average Freshness Hours:", round(average_freshness, 2))

# Step 8: Total row count
total_row_count = df["row_count"].sum()

print("Total Row Count:", total_row_count)

# Step 9: Total null count
total_null_count = df["null_count"].sum()

print("Total Null Count:", total_null_count)

# Step 10: Total duplicate count
total_duplicate_count = df["duplicate_count"].sum()

print("Total Duplicate Count:", total_duplicate_count)

# Step 11: Status distribution
status_summary = (
    df["status"]
    .value_counts()
    .reset_index()
)

status_summary.columns = ["status", "count"]

print("\nStatus Distribution:")
display(status_summary)

# Step 12: Alert rate
alert_count = (df["status"] == "ALERT").sum()

alert_rate = (alert_count / total_records) * 100

print("Alert Rate:", round(alert_rate, 2), "%")

# Step 13: Metric-wise analysis
metric_summary = (
    df.groupby("metric_name")
    .agg(
        total_value=("metric_value", "sum"),
        average_value=("metric_value", "mean"),
        record_count=("metric_name", "count"),
        average_freshness=("freshness_hours", "mean"),
        total_nulls=("null_count", "sum"),
        total_duplicates=("duplicate_count", "sum")
    )
    .reset_index()
)

print("\nMetric-wise Analysis:")
display(metric_summary)

# Step 14: Source-system analysis
source_summary = (
    df.groupby("source_system")
    .agg(
        total_value=("metric_value", "sum"),
        average_value=("metric_value", "mean"),
        records=("source_system", "count"),
        average_freshness=("freshness_hours", "mean"),
        total_nulls=("null_count", "sum"),
        total_duplicates=("duplicate_count", "sum")
    )
    .reset_index()
)

print("\nSource System Analysis:")
display(source_summary)

# Step 15: Status-wise analysis
status_analysis = (
    df.groupby("status")
    .agg(
        records=("status", "count"),
        average_freshness=("freshness_hours", "mean"),
        total_nulls=("null_count", "sum"),
        total_duplicates=("duplicate_count", "sum")
    )
    .reset_index()
)

print("\nStatus-wise Analysis:")
display(status_analysis)

# Step 16: Daily analysis
daily_summary = (
    df.groupby("event_date")
    .agg(
        metric_value=("metric_value", "sum"),
        average_freshness=("freshness_hours", "mean"),
        total_nulls=("null_count", "sum"),
        total_duplicates=("duplicate_count", "sum"),
        records=("event_id" if "event_id" in df.columns else "metric_name", "count")
    )
    .reset_index()
)

print("\nDaily Analysis:")
display(daily_summary.head())

# Step 17: Freshness analysis
freshness_summary = (
    df.groupby("source_system")["freshness_hours"]
    .agg(["mean", "min", "max"])
    .reset_index()
)

freshness_summary.columns = [
    "source_system",
    "average_freshness",
    "minimum_freshness",
    "maximum_freshness"
]

print("\nFreshness Analysis:")
display(freshness_summary)

# Step 18: Data-quality analysis
quality_summary = (
    df.groupby("source_system")
    .agg(
        total_records=("metric_name", "count"),
        total_nulls=("null_count", "sum"),
        total_duplicates=("duplicate_count", "sum"),
        average_freshness=("freshness_hours", "mean")
    )
    .reset_index()
)

print("\nData Quality Analysis:")
display(quality_summary)

# Step 19: High freshness records
high_freshness = df[df["freshness_hours"] > 8]

print("\nHigh Freshness Records:")
display(high_freshness.head())

# Step 20: Records with null issues
null_issue_records = df[df["null_count"] > 0]

print("\nRecords With Null Issues:")
display(null_issue_records.head())

# Step 21: Records with duplicate issues
duplicate_issue_records = df[df["duplicate_count"] > 0]

print("\nRecords With Duplicate Issues:")
display(duplicate_issue_records.head())

# Step 22: Alert records
alert_records = df[df["status"] == "ALERT"]

print("\nAlert Records:")
display(alert_records.head())

# Step 23: Top metrics by total value
top_metrics = (
    metric_summary
    .sort_values("total_value", ascending=False)
)

print("\nTop Metrics:")
display(top_metrics)

# Step 24: Source systems by metric value
top_sources = (
    source_summary
    .sort_values("total_value", ascending=False)
)

print("\nSource Systems by Metric Value:")
display(top_sources)

# Step 25: Overall business insights
print("\nBUSINESS INSIGHTS")

print(
    "1. Total records analyzed:",
    total_records
)

print(
    "2. Total metric value:",
    round(total_metric_value, 2)
)

print(
    "3. Average metric value:",
    round(average_metric_value, 2)
)

print(
    "4. Average freshness:",
    round(average_freshness, 2),
    "hours"
)

print(
    "5. Alert rate:",
    round(alert_rate, 2),
    "%"
)

print(
    "6. Total null count:",
    total_null_count
)

print(
    "7. Total duplicate count:",
    total_duplicate_count
)

# Step 26: Final analysis table
final_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Metric Value",
        "Average Metric Value",
        "Average Freshness Hours",
        "Alert Rate",
        "Total Null Count",
        "Total Duplicate Count"
    ],
    "Value": [
        total_records,
        round(total_metric_value, 2),
        round(average_metric_value, 2),
        round(average_freshness, 2),
        round(alert_rate, 2),
        total_null_count,
        total_duplicate_count
    ]
})

print("\nFinal Summary:")
display(final_summary)

# Save one analysis result
final_summary.to_excel("Python_Analysis_Result.xlsx", index=False)

print("\nAnalysis completed successfully.")
```
